## **ByteTrack Tracking and Violation Logic**

**CONCEPT: Detection vs Tracking — why you need both**

Detection answers: 'what objects are in THIS frame?'

Tracking answers: 'which object in frame 200 is the same person as in frame 150?'

Without tracking, you cannot measure duration. You'd see a 'NO-Hardhat' box in frame 200, another in frame 201, and have no way to know if they're the same person or two different people. Tracking assigns a persistent ID (track_id=3) to a person and keeps it as they move around. This lets you say: 'Person #3 has had no hardhat for 90 consecutive frames = 3 seconds.'That 3-second threshold separates a real violation from a momentary detection artifact.


**CONCEPT: How ByteTrack works internally**

Every frame: run YOLOv8 detection → get a list of boxes with confidence scores. ByteTrack takes ALL boxes (unlike older trackers that threw away low-confidence ones), and tries to match them to existing tracks using IoU (how much do boxes overlap) and a Kalman filter (predicts where a tracked object will be in the next frame based on velocity).
High-confidence boxes are matched first. Low-confidence boxes are then used to re-identify occluded tracks — this is ByteTrack's key innovation, reducing ID switches. New boxes that don't match any existing track get new IDs. Lost tracks are kept for N frames then deleted. The result: every detected person has a consistent integer ID.

### Violation detection logic

```python
# Cell 3 — define the violation detection system
import yaml
with open(r'C:\Users\user\Desktop\ML & DL projects\CV Real time Safety system\dataset\Construction Site Safety.v27-yolov8.yolov8\data.yaml') as f:
    class_names = yaml.safe_load(f)['names']
# Map class index → violation label for the two classes we care about
# Check YOUR data.yaml — indices may differ
VIOLATION_CLASSES = {}
for i, name in enumerate(class_names):
    if name in ('NO-Hardhat', 'NO-Safety Vest'):
        VIOLATION_CLASSES[i] = name
print('Violation class indices:', VIOLATION_CLASSES)
# This dict tracks when each (track_id, violation_type) pair was first seen
# Key: (track_id, violation_name) Value: frame_number it started
violation_timers = {}
# A violation is logged after this many consecutive frames of non-compliance
# At 30 FPS: 90 frames = 3 seconds
VIOLATION_THRESHOLD_FRAMES = 90
def process_frame(frame_idx, boxes_xyxy, track_ids, class_indices, fps=30.0):
    """
    Process one frame worth of detections.
    Returns list of violation dicts that crossed the threshold this frame.
    """
    confirmed_violations = []
    active_keys_this_frame = set()
    if track_ids is None:
        return confirmed_violations
    for box, track_id, cls in zip(boxes_xyxy, track_ids, class_indices):
        track_id = int(track_id)
        cls = int(cls)
        if cls in VIOLATION_CLASSES:
            viol_name = VIOLATION_CLASSES[cls]
            key = (track_id, viol_name)
            active_keys_this_frame.add(key)
        if key not in violation_timers:
    # First frame this person is violating — start the timer
            violation_timers[key] = frame_idx
        else:
            # Already tracking — check if threshold crossed
            frames_elapsed = frame_idx - violation_timers[key]
            if frames_elapsed >= VIOLATION_THRESHOLD_FRAMES:
                # Only log once per crossing (then reset timer)
                duration_seconds = frames_elapsed / fps
                x1, y1, x2, y2 = box.tolist()
                confirmed_violations.append({
                'track_id': track_id,
                'violation_type': viol_name,
                'duration_seconds': duration_seconds,
                'bbox': [x1, y1, x2, y2],
                'frame': frame_idx,
                })
                # Reset timer so we don't log every frame after threshold
                violation_timers[key] = frame_idx
    # Clear timers for people who are now compliant (not seen violating this frame)
    keys_to_remove = [k for k in violation_timers if k not in active_keys_this_frame
    and k[0] in [int(t) for t in (track_ids if track_ids is not None else [])]
    ]
    for k in keys_to_remove:
        del violation_timers[k]
    return confirmed_violations
```

```python
# run the full pipeline on the test video
import cv2
# Reset state
violation_timers.clear()
model_v = YOLO(r'C:\Users\user\Desktop\ML & DL projects\CV Real time Safety system\notebooks\runs\detect\runs\ppe\v1\weights\best.pt')
all_violations = []
cap = cv2.VideoCapture('test_video.mp4')
fps = cap.get(cv2.CAP_PROP_FPS) or 30.0
cap.release()
# track() is a generator — iterate over results frame by frame
for frame_idx, r in enumerate(model_v.track(
    source=r'C:\Users\user\Desktop\ML & DL projects\CV Real time Safety system\test_videos\construction_worker5.mp4',
    tracker='bytetrack.yaml',
    conf=0.4, iou=0.5, persist=True, stream=True, # stream=True = memory efficient
    )):
        violations = process_frame(
            frame_idx,
            r.boxes.xyxy if r.boxes.id is not None else None,
            r.boxes.id if r.boxes.id is not None else None,
            r.boxes.cls if r.boxes.id is not None else None,
            fps=fps
        )
        all_violations.extend(violations)
print(f'\nTotal violations logged: {len(all_violations)}')
for v in all_violations:
    print(f" Frame {v['frame']:4d} | Track {v['track_id']:3d} | {v['violation_type']:20s} | {v['duration_seconds']:.1f}s")
```
